## Problem 1

Using  the [GaussianNB](https://scikit-learn.org/stable/modules/generated/sklearn.naive_bayes.GaussianNB.html) function construct the  Gaussian  Bayes classifier using only one feature. Which feature will you choose? Calculate the confusion matrix (normalized as to show rates), ROC AUC score and plot ROC curve. Do this both for training and validation set. Plot both curves on the same plot.  

```python
from sklearn.naive_bayes import GaussianNB
```

__Hint__ For calculating metrics and plotting ROC curves you may use functions from scikit-learn: `roc_curve`, `roc_auc_score` and `confusion matrix`. For estimating normal distribution parameters  use `norm.fit` `from scipy.stats`. Use `norm.pdf` for normal probability density function.

```python
from sklearn.metrics import roc_auc_score, roc_curve, confusion_matrix
```

Which feature did you choose?

```python
# Problem 1: GaussianNB with one feature
# Choose feature 'a0' - it shows the best separation between classes in the histograms

gnb1 = GaussianNB()
gnb1.fit(data_train[['a0']], lbls_train)

# Training set
proba_train_a0 = gnb1.predict_proba(data_train[['a0']])[:, 1]
pred_train_a0 = gnb1.predict(data_train[['a0']])

# Test set
proba_test_a0 = gnb1.predict_proba(data_test[['a0']])[:, 1]
pred_test_a0 = gnb1.predict(data_test[['a0']])

# Confusion matrices (normalized)
print("Training set confusion matrix (normalized):")
print(confusion_matrix(lbls_train, pred_train_a0, normalize='true'))
print("\nTest set confusion matrix (normalized):")
print(confusion_matrix(lbls_test, pred_test_a0, normalize='true'))

# AUC scores
auc_train_a0 = roc_auc_score(lbls_train, proba_train_a0)
auc_test_a0 = roc_auc_score(lbls_test, proba_test_a0)
print(f"\nAUC train: {auc_train_a0:.4f}")
print(f"AUC test:  {auc_test_a0:.4f}")

# ROC curves
fprs_train_a0, tprs_train_a0, _ = roc_curve(lbls_train, proba_train_a0)
fprs_test_a0, tprs_test_a0, _ = roc_curve(lbls_test, proba_test_a0)

fig, ax = plt.subplots(figsize=(8, 8))
ax.set_aspect(1.0)
ax.plot(fprs_train_a0, tprs_train_a0, label=f'Train AUC={auc_train_a0:.4f}')
ax.plot(fprs_test_a0, tprs_test_a0, label=f'Test AUC={auc_test_a0:.4f}')
ax.plot([0, 1], [0, 1], 'k--')
ax.set_xlabel('FPR')
ax.set_ylabel('TPR')
ax.set_title('Problem 1: GaussianNB with feature a0')
ax.legend()
plt.show()
```
